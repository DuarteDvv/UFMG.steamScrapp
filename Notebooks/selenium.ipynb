```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import pandas as pd
from datetime import datetime, timedelta
```

```python
# Configuração das opções do navegador
options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/89.0.4389.82 Safari/537.36")

# Inicializa o navegador com as opções configuradas
navegador = webdriver.Chrome(options=options)

# URL da página a ser acessada
url = "https://steamdb.info/charts/?all"
navegador.get(url)
```

```python
# Localiza a tabela de dados na página
tabela = navegador.find_element(By.CLASS_NAME, "dataTable")

# Localiza todos os elementos que representam apps na tabela
listaApps = tabela.find_elements(By.CLASS_NAME, "app")

# Lista para armazenar os dados extraídos
lista = []

# Lista para armazenar os tempos de execução de cada iteração
listaTempos = []
qtdFaltando = len(listaApps)

# Itera sobre cada app na lista de apps
for app in listaApps:

    start = datetime.now()

    # num = int(app.find_element(By.CLASS_NAME, "dt-type-numeric").text.replace(".", ""))

    # Extrai os elementos de tipo numérico
    aux = app.find_elements(By.CLASS_NAME, "dt-type-numeric")


    # Extrai o nome do app
    aux2 = app.find_elements(By.CSS_SELECTOR, "td")
    nome = aux2[2].text

    # Extrai e converte os dados de jogadores atuais, pico de 24h e pico de todos os tempos
    num = int(aux[0].text.replace(".", ""))
    currentPlayers = int(aux[2].text.replace(",", ""))
    peak24h = int(aux[3].text.replace(",", ""))
    allTimePeak = int(aux[4].text.replace(",", ""))

    # Extrai o ID do jogo a partir da URL
    gameID = aux2[2].find_element(By.TAG_NAME, "a").get_attribute("href")
    gameID = gameID.replace("https://steamdb.info/app/", "").replace("/charts/", "")
    gameID = int(gameID)

    # Adiciona os dados extraídos à lista
    lista.append([num, nome, currentPlayers, peak24h, allTimePeak, gameID])

    end = datetime.now()


    # print(num, nome, currentPlayers, peak24h, allTimePeak)

    qtdFaltando-=1
    listaTempos.append((end - start).total_seconds())

    # A cada 100 iterações, calcula o tempo médio e estima o horário de término
    if(num%100 == 0):
        mediaSegundos = sum(listaTempos) / len(listaTempos)
        hora = datetime.now() + timedelta(seconds=mediaSegundos*qtdFaltando)
        stringHora = hora.strftime("%H:%M:%S")
        print(f"{num}: Horário esperado de término: {stringHora}")

    # break

# Converte a lista de dados para um DataFrame do pandas
df = pd.DataFrame(lista, columns=["num", "nome", "currentPlayers", "peak24h", "allTimePeak", "gameID"])
df
```
